```python
# Conversation Analyzer

import re
from collections import Counter

conversation = """
Alice: Hey! How are you doing today?
Bob: I am doing great! I completed my project.
Alice: That's amazing. Are you coming to college tomorrow?
Bob: Yes, I will come in the morning.
Alice: Great! We can discuss the project then.
Bob: Sure, that sounds good!
"""

# Extract messages
messages = re.findall(r'(\w+):\s*(.*)', conversation)

speakers = [speaker for speaker, message in messages]
texts = [message for speaker, message in messages]

# Count messages
message_count = Counter(speakers)

# Word analysis
all_words = []

for text in texts:
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
    all_words.extend(words)

stop_words = {
    "the", "is", "am", "are", "a", "an", "i", "you",
    "to", "in", "of", "and", "it", "we", "can", "will",
    "that", "this", "how", "do", "my", "on", "then"
}

meaningful_words = [
    word for word in all_words
    if word not in stop_words
]

word_frequency = Counter(meaningful_words)

# Sentiment words
positive_words = {
    "great", "amazing", "good", "happy", "excellent",
    "nice", "love", "sure"
}

negative_words = {
    "bad", "sad", "angry", "hate", "terrible",
    "problem", "worried"
}

positive_count = sum(
    1 for word in all_words if word in positive_words
)

negative_count = sum(
    1 for word in all_words if word in negative_words
)

# Questions
question_count = conversation.count("?")

# Sentiment
if positive_count > negative_count:
    sentiment = "Positive"
elif negative_count > positive_count:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

# Most active speaker
most_active = message_count.most_common(1)[0]

print("=" * 55)
print("              CONVERSATION ANALYZER")
print("=" * 55)

print("\nTotal Messages :", len(messages))
print("Participants   :", ", ".join(message_count.keys()))
print("Questions      :", question_count)

print("\nMessage Count:")
for speaker, count in message_count.items():
    print(f"  {speaker}: {count}")

print("\nMost Active Speaker:")
print(f"  {most_active[0]} ({most_active[1]} messages)")

print("\nMost Used Words:")
for word, count in word_frequency.most_common(5):
    print(f"  {word}: {count}")

print("\nConversation Sentiment:")
print("  Positive Words :", positive_count)
print("  Negative Words :", negative_count)
print("  Overall Tone   :", sentiment)

print("\n" + "=" * 55)
print("Conversation Analysis Completed!")
print("=" * 55)
```
